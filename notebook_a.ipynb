```python
import jieba
import jieba.analyse

# 分词示例
text = "我爱自然语言处理"
print("精确模式：")
print("/ ".join(jieba.cut(text, cut_all=False)))

print("全模式：")
print("/ ".join(jieba.cut(text, cut_all=True)))

# 关键词提取示例
text_for_keywords = "我爱自然语言处理。自然语言处理是计算机科学和人工智能领域的一个重要方向。"
print("关键词提取：")
print(jieba.analyse.extract_tags(text_for_keywords, topK=5))
```
